# BERT (Bidirectional Encoder Representations from Transformers)

## 마스크드 언어 모델과 토크나이저

```python
# Hugging Face의 transformers 라이브러리에서 필요한 클래스 임포트
# BertForMaskedLM: [MASK] 토큰을 예측하기 위한 사전 학습된 BERT 모델
# AutoTokenizer: 모델 이름에 맞는 토크나이저를 자동으로 불러오는 클래스
from transformers import BertForMaskedLM
from transformers import AutoTokenizer
```

```python
# 'bert-large-uncased' 사전 학습된 모델과 해당 토크나이저 로드
# bert-large-uncased: 대문자를 구분하지 않는(uncased) 대형 BERT 모델
model = BertForMaskedLM.from_pretrained('bert-large-uncased')
tokenizer = AutoTokenizer.from_pretrained("bert-large-uncased")
```

```python
### [MASK]의 위치에 해당하는 단어를 예측
## 정수 인코딩 결과
inputs = tokenizer('good [MASK].')
print(inputs)

inputs = tokenizer('Soccer is a really fun [MASK].')
print(inputs['input_ids'])
```

```python
### 문장을 구분하는 세그먼트 인코딩 결과
# 단일 문장 입력 시 모두 0
print(inputs['token_type_ids'])
```

```python
# 패딩 영역이 아닌 실제 의미 있는 토큰(1)인지 나타냄
### attention_mask
print(inputs['attention_mask'])
```

### [MASK] 토큰 예측하기

```python
from transformers import FillMaskPipeline
pip = FillMaskPipeline(model=model, tokenizer=tokenizer) # 파이프라인 제공
```

```python
pip('Soccer is a really fun [MASK].')
```

```python
pip('The Avengers is a really fun [MASK].')
```

```python
pip('I went to [MASK] this morning.')
```

## BERT를 이용한 다음 문장 예측

```python
import torch
from transformers import BertForNextSentencePrediction
from transformers import AutoTokenizer
```

```python
model = BertForNextSentencePrediction.from_pretrained('bert-base-uncased')
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
```

```python
prompt = "In Italy, pizza served in formal settings, such as at a restaurant, is presented unsliced."
next_sentence = "pizza is eaten with the use of a knife and fork. In casual settings, however, it is cut into wedges to be eaten while held in the hand."
```

```python
encoding = tokenizer(prompt, next_sentence, return_tensors='pt')
```

```python
## 정수 인코딩
print(encoding['input_ids'])
```

```python
print(tokenizer.cls_token, ':', tokenizer.cls_token_id)
print(tokenizer.sep_token, ':' , tokenizer.sep_token_id)
```

```python
print(tokenizer.decode(encoding['input_ids'][0]))
```

```python
print(encoding['token_type_ids'])
```

### 다음 문장 예측하기

```python
pred = model(encoding['input_ids'], token_type_ids=encoding['token_type_ids'])
probs = torch.nn.functional.softmax(pred.logits, dim=1)  # Softmax 적용하여 확률 얻기
print(probs)
```

```python
# 상관없는 두 개의 문장
prompt = "In Italy, pizza served in formal settings, such as at a restaurant, is presented unsliced."
next_sentence = "The sky is blue due to the shorter wavelength of blue light."
encoding = tokenizer(prompt, next_sentence, return_tensors='pt')

pred = model(encoding['input_ids'], token_type_ids=encoding['token_type_ids'])
probs = torch.nn.functional.softmax(pred.logits, dim=1)  # Softmax 적용하여 확률 얻기
next_sentence_label = torch.argmax(probs, dim=1).item()  # 예측된 라벨 얻기
print('최종 예측 레이블 :', next_sentence_label)
```

### 한국어 모델의 다음 문장 예측 모델과 토크나이저

```python
import torch
from transformers import BertForNextSentencePrediction
from transformers import AutoTokenizer
```

```python
model = BertForNextSentencePrediction.from_pretrained('klue/bert-base')
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base")
```

```python
# 이어지는 두 개의 문장
prompt = "2002년 월드컵 축구대회는 일본과 공동으로 개최되었던 세계적인 큰 잔치입니다."
next_sentence = "여행을 가보니 한국의 2002년 월드컵 축구대회의 준비는 완벽했습니다."
encoding = tokenizer(prompt, next_sentence, return_tensors='pt')

pred = model(encoding['input_ids'], token_type_ids=encoding['token_type_ids'])
probs = torch.nn.functional.softmax(pred.logits, dim=1)  # Softmax 적용하여 확률 얻기
next_sentence_label = torch.argmax(probs, dim=1).item()  # 예측된 라벨 얻기
print('최종 예측 레이블 :', next_sentence_label)
```

```python
# 상관없는 두 개의 문장
prompt = "2002년 월드컵 축구대회는 일본과 공동으로 개최되었던 세계적인 큰 잔치입니다."
next_sentence = "극장가서 로맨스 영화를 보고싶어요"
encoding = tokenizer(prompt, next_sentence, return_tensors='pt')

pred = model(encoding['input_ids'], token_type_ids=encoding['token_type_ids'])
probs = torch.nn.functional.softmax(pred.logits, dim=1)  # Softmax 적용하여 확률 얻기
next_sentence_label = torch.argmax(probs, dim=1).item()  # 예측된 라벨 얻기
print('최종 예측 레이블 :', next_sentence_label)
```
